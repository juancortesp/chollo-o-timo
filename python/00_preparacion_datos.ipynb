```python
# ── Celda 1 · Librerías ─────────────────────────────
import pandas as pd
from pathlib import Path

# ── Celda 2 · Rutas ─────────────────────────────────
CARPETA = Path(r"C:\Users\jhonn\Desktop\NEOLAND\PROYECTO - COCHE\data_raw")
RUTA_ENTRADA = CARPETA / "european_dataset.xlsx"
RUTA_SALIDA = CARPETA / "coches_es_2026.xlsx"

print("¿Existe el archivo?", RUTA_ENTRADA.exists())

# Ver qué archivos hay en la carpeta
for archivo in CARPETA.iterdir():
    print(archivo.name)
    
# ── Celda 3 · Leer el archivo ───────────────────────
# Primero intenta leerlo como parquet; si falla, como Excel
try:
    df = pd.read_parquet(RUTA_ENTRADA)
    print("Leído como parquet")
except Exception:
    df = pd.read_excel(RUTA_ENTRADA)
    print("Leído como Excel")

print("Filas y columnas totales:", df.shape)
print(df.columns.tolist())

# ── Celda 4 · Ver cómo viene codificado el país ─────
print(df["seller_country"].value_counts())

# ── Celda 5 · Filtrar España ────────────────────────
# Si en la celda 4 España aparece con otro texto, añádelo a esta lista
VALORES_ESPANA = ["ES", "Spain", "España", "es", "ESP"]

df_es = df[df["seller_country"].isin(VALORES_ESPANA)].copy()
print("Filas España:", df_es.shape)

# ── Celda 6 · Guardar en Excel ──────────────────────
assert len(df_es) > 0, "No hay filas de España: revisa VALORES_ESPANA con la celda 4"

df_es.to_excel(RUTA_SALIDA, index=False)
print("Guardado en:", RUTA_SALIDA)
```
